```python
posicao_erlenmeyer = (243,171,240)
posicao_caixa_ponteiras = [144.9,312,160]

acido_cloroaurico = (31,191,168)
nitrato_prata = (31,178,171)
acido_ascorbico = (31,165,171)
borohidreto = (31,155,171)
solucao_previa = (31,143,171)


#ctab = (133,133,160)

```

```python
agitador.ligar()


#Acoplar Primeira Ponteira
pipeta.attach_tip(tip_box_position=posicao_caixa_ponteiras,safe_height=altura_segura,velocity=velocidade_movimento_xy)


#Pipeta 1000ul de ácido cloroáurico
pipeta.pipette_liquid(start_position_xyz=acido_cloroaurico, end_position_xyz=posicao_erlenmeyer, volume_ul=1000, safe_height=altura_segura, dispense_velocity=300)

#Remover ponteira
pipeta.discart_tip(discard_position=posicao_descarte,safe_height=altura_segura,velocity=velocidade_movimento_xy)


#Acoplar Segunda Ponteira
pipeta.attach_tip(tip_box_position=posicao_caixa_ponteiras,safe_height=altura_segura,velocity=velocidade_movimento_xy)

#Pipeta 200ul de nitrato de prata
pipeta.pipette_liquid(start_position_xyz=nitrato_prata, end_position_xyz=posicao_erlenmeyer, volume_ul=200, safe_height=altura_segura, dispense_velocity=300)

#Remover ponteira
pipeta.discart_tip(discard_position=posicao_descarte,safe_height=altura_segura,velocity=velocidade_movimento_xy)


#Acoplar Terceiro Ponteira
pipeta.attach_tip(tip_box_position=posicao_caixa_ponteiras,safe_height=altura_segura,velocity=velocidade_movimento_xy)

#Pipeta 120ul de nitrato de prata
pipeta.pipette_liquid(start_position_xyz=nitrato_prata, end_position_xyz=posicao_erlenmeyer, volume_ul=120, safe_height=altura_segura, dispense_velocity=300)

#Remover ponteira
pipeta.discart_tip(discard_position=posicao_descarte,safe_height=altura_segura,velocity=velocidade_movimento_xy)


#Acoplar Quarta Ponteira
pipeta.attach_tip(tip_box_position=posicao_caixa_ponteiras,safe_height=altura_segura,velocity=velocidade_movimento_xy)

#Pipeta 100ul de seeds de ouro
pipeta.pipette_liquid(start_position_xyz=solucao_previa, end_position_xyz=posicao_erlenmeyer, volume_ul=100, safe_height=altura_segura, dispense_velocity=500)

#Remover ponteira
pipeta.discart_tip(discard_position=posicao_descarte,safe_height=altura_segura,velocity=velocidade_movimento_xy)

time.sleep(60*25)

#Desativar Agitador
agitador.desligar()
agitador.fechar()
```
