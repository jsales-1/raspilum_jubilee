```python
import os
import time

for i in range(1, 101):

    nome_amostra = f"dados_{i*100}_ul.txt"

    pipeta.pipette_liquid(posicao_leite,posicao_agua,100,safe_height=altura_segura,velocidade_dispensacao=300)
    pipeta.pipette_liquid(posicao_leite,posicao_validacao,100,safe_height=altura_segura,velocidade_dispensacao=300)

    agitador.ligar()
    time.sleep(15)
    agitador.desligar()
    time.sleep(15)
    sensor.save_readings(
        filename=f"Teste_agua_pura/{nome_amostra}",
        n=10,
        interval=0.001
    )

    while nome_amostra not in os.listdir("Teste_agua_pura"):
        time.sleep(1)

    print(f"{nome_amostra} salvo!")
```
